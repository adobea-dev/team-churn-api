```python
#eda
from ucimlrepo import fetch_ucirepo
import pandas as pd

dataset = fetch_ucirepo(id = 563)

X = dataset.data.features.copy()
y = dataset.data.targets.iloc[ :, 0]

df = X.copy()
df["Churn"] = y

print(df.shape)
df.head()

```

```python
#loading the saved model
import joblib
from pathlib import Path

model_path = Path("models/churn_pipeline.joblib")

model = joblib.load(model_path)

print("Model loaded successfully!")
print(model)
```

```python
#making a prediction
import pandas as pd

sample_customer_prediction = X.iloc[[0]].copy()

#standardize column names as done in the training
sample_customer_prediction.columns = [
    " ".join(col.split()) for col in sample_customer_prediction.columns
]

prediction = model.predict(sample_customer_prediction)

# Predict probability of churn
probability = model.predict_proba(sample_customer_prediction)

print("Prediction:", prediction[0])
print("Churn probability:", probability[0][1])
```
